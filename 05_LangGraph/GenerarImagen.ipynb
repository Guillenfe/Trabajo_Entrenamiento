```python
# Interfaz Gradio para generar imágenes con Stable Diffusion
import torch
import gradio as gr
from diffusers import StableDiffusionPipeline
from huggingface_hub import login
from google.colab import userdata

hf_token = userdata.get("ApiKeyHugginFace")
login(token=hf_token)

device = "cuda" if torch.cuda.is_available() else "cpu"

MODELS = {
    "OpenJourney": "prompthero/openjourney",
}

pipes = {}

def load_model(model_name):
    if model_name not in pipes:
        pipe = StableDiffusionPipeline.from_pretrained(
            MODELS[model_name],
            torch_dtype=torch.float16
        ).to(device)
        pipe.safety_checker = None
        pipes[model_name] = pipe
    return pipes[model_name]

def generate(prompt, steps, guidance, model_name):
    pipe = load_model(model_name)
    image = pipe(
        prompt,
        num_inference_steps=steps,
        guidance_scale=guidance
    ).images[0]
    return image

with gr.Blocks() as demo:
    gr.Markdown("## Generador de pósters cinematográficos")

    model_selector = gr.Dropdown(
        choices=list(MODELS.keys()),
        value="OpenJourney",
        label="Modelo"
    )

    prompt = gr.Textbox(
        label="Prompt",
        placeholder="Official cinematic poster of the movie..."
    )

    steps = gr.Slider(20, 80, value=60, step=1, label="Inference steps")
    guidance = gr.Slider(1, 15, value=7.5, step=0.5, label="Guidance scale")

    generate_btn = gr.Button("Generar")
    output = gr.Image(label="Resultado")

    generate_btn.click(
        generate,
        inputs=[prompt, steps, guidance, model_selector],
        outputs=output
    )

demo.launch()
```
