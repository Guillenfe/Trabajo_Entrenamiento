Podemos mejorar la clase Website para que incorpore información de los `links`que tenga la página que queremos resumir:

```python
import os
import json
load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")

openai = OpenAI(
    base_url="https://generativelanguage.googleapis.com/v1beta",
    api_key=api_key
)

# Crear un archivo de ejemplo si no existe
if not os.path.exists("data_film.json"):
    sample_data = {
        "title": "El Padrino",
        "rating": 9.2,
        "synopsis": "La historia de la familia mafiosa Corleone en Nueva York.",
        "cast": ["Marlon Brando", "Al Pacino", "James Caan"]
    }
    with open("data_film.json", "w", encoding="utf-8") as f:
        json.dump(sample_data, f, ensure_ascii=False, indent=2)

with open("data_film.json", "r", encoding="utf-8") as f:
    film_data = json.load(f)

prompt = f"""
Eres un crítico de cine. Resume la información de esta película:
Título: {film_data['title']}
Valoración media: {film_data['rating']}
Sinopsis: {film_data['synopsis']}
Reparto: {', '.join(film_data['cast'])}

Por favor, genera un análisis breve en español en formato markdown.
"""

response = openai.chat.completions.create(
    model="gemini-2.0-flash",
    messages=[
        {"role": "system", "content": "Eres un crítico de cine experto y preciso."},
        {"role": "user", "content": prompt}
    ]
)

print(response.choices[0].message.content)
```
