# Scraping de reseñas IMDb con RapidAPI

Este cuaderno explica paso a paso cómo funciona el script `imdb_rapidapi_reviews_env.py`.

```python
import os
from pathlib import Path
from dotenv import load_dotenv
import requests
import json
from datetime import datetime, timezone
```

## 1️⃣ Cargar variables de entorno
El script original usa `.env` para guardar la API key y host de RapidAPI.

```python
# Cargar .env
env_path = Path('.') / '.env'
load_dotenv(dotenv_path=env_path)

RAPIDAPI_KEY = os.getenv('apiOMDb')  # tu variable se llama apiOMDb
RAPIDAPI_HOST = os.getenv('RAPIDAPI_HOST')  # RapidAPI host

print('RapidAPI key cargada:', bool(RAPIDAPI_KEY))
```

## 2️⃣ Función para llamar a RapidAPI
Esta función hace la llamada HTTP y devuelve el JSON resultante.

```python
def call_rapidapi(path: str, params=None):
    url = f"https://{RAPIDAPI_HOST}{path}"
    headers = {
        'x-rapidapi-key': RAPIDAPI_KEY,
        'x-rapidapi-host': RAPIDAPI_HOST
    }
    resp = requests.get(url, headers=headers, params=params)
    return resp.json()
```

## 3️⃣ Llamada de prueba
Usamos el mismo endpoint que en tu script original.

```python
imdb_id = 'tt0110912'  # Pulp Fiction
path = '/title/get-user-reviews'
params = {'tconst': imdb_id}

data = call_rapidapi(path, params)
print(json.dumps(data[:2], indent=2))  # mostramos las primeras 2 reseñas
```

## 4️⃣ Guardar reseñas en JSON

```python
output_file = f'{imdb_id}_reviews_example.json'
with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print(f'Guardadas {len(data)} reseñas en {output_file}')
```
